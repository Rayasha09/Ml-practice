## Основные наблюдения

1. Больше всего заказов приходится на субботу, а меньше всего — на пятницу.
2. Средний счёт в ресторане составляет примерно 19,79 доллара.
3. В датасете есть числовые признаки (`total_bill`, `tip`, `size`) и категориальные признаки (`sex`, `smoker`, `day`, `time`).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline

sns.set_theme(style="whitegrid")

pd.set_option("display.max_columns", None)
```

```python
df = sns.load_dataset("tips")
df.head()
```

```python
print(f"Размер таблицы (строки, столбцы): {df.shape}")
```

```python
print("Типы столбцов:")
print(df.dtypes)
```

```python
df.isna().sum()
```

```python
df.describe(include="all")
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df["total_bill"], bins=20, kde=True)

plt.title("Распределение суммы счёта (total_bill)")
plt.xlabel("Сумма счёта, $")
plt.ylabel("Количество наблюдений")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="day",
    order=["Thur", "Fri", "Sat", "Sun"]
)

plt.title("Количество заказов по дням недели")
plt.xlabel("День недели")
plt.ylabel("Количество заказов")

plt.show()
```
